```python
import pandas as pd

sales = pd.read_csv("../data/raw/sales_transactions.csv")
customers = pd.read_csv("../data/raw/customers.csv")
products = pd.read_csv("../data/raw/products.csv")
inventory = pd.read_csv("../data/raw/inventory.csv")
returns = pd.read_csv("../data/raw/returns.csv")

sales.head()
```

```python
print("SALES DATA")
print(sales.head())

print("\nCUSTOMERS DATA")
print(customers.head())

print("\nPRODUCTS DATA")
print(products.head())

print("\nINVENTORY DATA")
print(inventory.head())

print("\nRETURNS DATA")
print(returns.head())
```

```python
print("Sales Columns")
print(sales.columns)

print("\nCustomer Columns")
print(customers.columns)

print("\nProduct Columns")
print(products.columns)

print("\nInventory Columns")
print(inventory.columns)

print("\nReturns Columns")
print(returns.columns)
```

```python
print(sales.info())
print(customers.info())
print(products.info())
print(inventory.info())
print(returns.info())
```

```python
print(sales.isnull().sum())
print(customers.isnull().sum())
print(products.isnull().sum())
print(inventory.isnull().sum())
print(returns.isnull().sum())
```

```python
sales.drop_duplicates(inplace=True)
customers.drop_duplicates(inplace=True)
products.drop_duplicates(inplace=True)
inventory.drop_duplicates(inplace=True)
returns.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully")
```

```python
sales.dropna(inplace=True)
customers.dropna(inplace=True)
products.dropna(inplace=True)
inventory.dropna(inplace=True)
returns.dropna(inplace=True)

print("Missing Values Removed")
```

```python
sales['Date'] = pd.to_datetime(sales['Date'])

print(sales.dtypes)
```

```python
merged_df = sales.merge(
    customers,
    on='Customer_ID'
)

print(merged_df.head())
```

```python
merged_df = merged_df.merge(
    products,
    on='Product_ID'
)

print(merged_df.head())
```

```python
merged_df['Month'] = merged_df['Date'].dt.month
merged_df['Year'] = merged_df['Date'].dt.year
merged_df['Day'] = merged_df['Date'].dt.day_name()

print(
    merged_df[['Date', 'Month', 'Year', 'Day']].head()
)
```

```python
total_revenue = merged_df['Total_Sales'].sum()

print("Total Revenue =", total_revenue)
```

```python
total_orders = merged_df['Order_ID'].count()

print("Total Orders =", total_orders)
```

```python
category_sales = merged_df.groupby(
    'Category'
)['Total_Sales'].sum()

print(category_sales)
```

```python
channel_sales = merged_df.groupby(
    'Channel'
)['Total_Sales'].sum()

print(channel_sales)
```

```python
city_sales = merged_df.groupby(
    'City'
)['Total_Sales'].sum()

print(city_sales.sort_values(ascending=False))
```

```python
top_products = merged_df.groupby(
    'Product_Name'
)['Total_Sales'].sum()

print(
    top_products.sort_values(
        ascending=False
    ).head(10)
)
```
